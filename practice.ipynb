```python
playlist = ['song1','song2','song3','song4','song5']

for song in playlist:
    print(song)
```

```python
for i in range (10):
    print(i)
```

```python
for i in range (1,6):     #1 is inclusive and #6 is exclusive
    print(i)
```

```python
#nested Loop
for i in range(2,5):
    for j in range(1,4):
        print(f"{i} x {j} = {i*j}")
    print("=="*10)
```

#while loop

syntax:

#while condition:
  #code execute till the condition satisfy

```python
count=0
while count<5:
    print("Count:",count)
    count+=1
print("Loop Finished")    
```

```python
password=''
while password!="admin@123":
    password=input("Enter the password")
print("Permission Granted")
```

```python
count =0

while count<10:
    if count ==3:
        break
    print("Count:",count)
    count+=1
print("Loop finished")    
    
```
